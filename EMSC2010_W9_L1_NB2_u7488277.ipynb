# EMSC2010-W9-L1-NB2

#### The logistic population model:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size
* $x_{\mathrm{next}}$ is next year's population size
* $a$ is the ability to reproduce.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### Code the logistic population model
```

```python
# Calculate x_next
# Define constants
x_initial = 0.5
a = 2.5
x = np.zeros(50)
x[0] = x_initial

for i in range(1,50):
    x[i] = a * x[i-1] * (1 - x[i-1])

plt.plot(x)
plt.xlabel('Year')
plt.ylabel('Population')
plt.ylim(0,1)
plt.show()
```

#### The logistic population model with harvesting:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}}) - h$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $h$ is the size of the harvest between 0 and 1.


```python
### Code the logistic population model with harvesting
```

```python
# Calculate x_next
# Define constants
x_initial = 0.5
a = 2.5
h = 0.1
x = np.zeros(50)
x[0] = x_initial

for i in range(1,50):
    x[i] = a * x[i-1] * (1 - x[i-1]) - h
    x[i] = np.maximum(x[i],0)

plt.plot(x)
plt.xlabel('Year')
plt.ylabel('Population')
plt.ylim(0,1)
plt.show()
```

#### The logistic population model with randomness:

$x_{\mathrm{next}} = (a+r) x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $r$ is a normally distributed random number that changes each year.

```python
### Code the logistic population model with randomness
```

```python
# Calculate x_next
# Define constants
x_initial = 0.5
a = 2.5
x = np.zeros(50)
x[0] = x_initial

for i in range(1,50):
    r = np.random.randn()*0.6 # Multiply by whatever you want your standard distribution to be
    x[i] = (a + r) * x[i-1] * (1 - x[i-1])
    x[i] = np.maximum(x[i],0)

plt.plot(x)
plt.xlabel('Year')
plt.ylabel('Population')
plt.ylim(0,1)
plt.show()
```

#### The logistic map:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size
* $x_{\mathrm{next}}$ is next year's population size
* $a$ is the ability to reproduce.

Follow these steps:

1. Choose a random value of $a$ from 0 to 4.

2. Choose a random starting population x_{\mathrm{initial}} from 0 to 1

3. Run the model for 2,000 generations and record the $x$ value of the last generation.

4. Repeat steps 1-3 a large number of times (thousands)

5. Make a scatter plot of $a$ (x-axis) versus final $x$ (y-axis).



```python
### Code the logistic map
```

```python
# Calculate x_next
num_simulations = 25000 # Repeat steps 1-3 a large number of times

all_a_values = []
all_x_values = []
all_final_x_values = []

for j in range(num_simulations):
    a = np.random.uniform(0,4)
    x = np.zeros(2000)
    x[0] = np.random.rand()

    for i in range(1,2000):
        x[i] = a * x[i-1] * (1 - x[i-1])
        x[i] = np.maximum(x[i],0)

    all_a_values.append(a)
    all_x_values.append(x[0])
    all_final_x_values.append(x[-1])
```

```python
plt.figure(figsize=(10,5))
plt.scatter(all_a_values, all_final_x_values, s=1.5, alpha=0.5)
plt.xlabel('a value')
plt.ylabel('Final Population (x)')
plt.title('Logistic Map: (a vs. Final x )')
plt.ylim(0,1)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(all_x_values, all_final_x_values, s=1.5, alpha=0.5)
plt.xlabel('x value')
plt.ylabel('Final Population (x)')
plt.title('Logistic Map: (Initial x vs. Final x )')
plt.ylim(0,1)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
